```python
import numpy as np
import pandas as pd
```

## Part 1: NumPy Array Operations

### 1. Create a 1D array

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

### 2. Inspection and properties

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

### 3. Array operations

```python
# celsius to fahrenheit
temp_fahrenheit = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", temp_fahrenheit)
```

```python
print("Maximum:", temperatures_w1.max())
print("Minimum:", temperatures_w1.min())
print("Mean:", round(temperatures_w1.mean(), 2))
```

### 4. Slicing and indexing

```python
print("First three days:", temperatures_w1[:3])
print("Weekend (last two days):", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

### 5. Create a 2D array

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],   # week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]    # week 2
])
print(temperatures)
```

### 6. Inspect and slice the 2D array

```python
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total elements:", temperatures.size)
```

```python
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
print()
print("Weekend of both weeks:")
print(temperatures[:, -2:])
```

## Part 2: Pandas Series

### 1. Creating the Series

```python
marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

### 2. Indexing and slicing

```python
# 1st rank student using integer position
print("1st rank mark:", marks.values[0])
```

```python
# top 3 ranks using loc (labels)
marks.loc['Rank1':'Rank3']
```

```python
# 3rd rank student using iloc
print("3rd rank mark:", marks.iloc[2])
```

```python
# boolean mask, marks greater than 90
marks[marks > 90]
```

### 3. Manipulating the Series

```python
# change 1st rank mark to 100
marks['Rank1'] = 100
marks
```

```python
# remove the last rank
marks = marks.drop('Rank5')
marks
```

```python
# cgpa = marks / 10
cgpa = marks / 10
cgpa
```

## Part 3: Pandas DataFrame

### 1. Creating the DataFrame

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

### 2. Data exploration

```python
transactions.head()
```

```python
transactions.tail()
```

```python
print("Shape:", transactions.shape)
print("Columns:", list(transactions.columns))
print()
print(transactions.dtypes)
```

```python
transactions.info()
```

```python
# only ProductCategory and Amount
transactions[['ProductCategory', 'Amount']]
```

```python
# last 3 columns
transactions.iloc[:, -3:]
```

```python
# North region and amount more than 200
transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
```

```python
transactions['ProductCategory'].value_counts()
```

```python
transactions['Region'].unique()
```

```python
# average amount for each region
transactions.groupby('Region')['Amount'].mean()
```

### 3. Manipulating the DataFrame

```python
# change amount of transaction 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

```python
# new column Discount = 10% of Amount
transactions['Discount'] = transactions['Amount'] * 0.10
transactions
```

```python
# remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
transactions
```

```python
# delete the Discount column
transactions = transactions.drop(columns='Discount')
transactions
```
